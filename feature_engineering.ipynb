```python
import pandas as pd
from IPython.display import display

# ---- Step 0: Load the dataset (already in your Colab files) ----
df = pd.read_csv('Pokemon.csv')

print("Original data (first 10 rows):")
display(df.head(10))


# ===================================================================
# 1. INTERACTION FEATURES
# Combine two numeric columns to create a new feature that captures
# their combined effect. Example: Attack * Defense = "Battle Power",
# and Sp. Atk * Sp. Def = "Special Power".
# ===================================================================
df['Attack_Defense_Interaction'] = df['Attack'] * df['Defense']
df['SpAtk_SpDef_Interaction'] = df['Sp. Atk'] * df['Sp. Def']

print("\nAfter adding interaction features:")
display(df[['Name', 'Attack', 'Defense', 'Attack_Defense_Interaction',
            'Sp. Atk', 'Sp. Def', 'SpAtk_SpDef_Interaction']].head(10))


# ===================================================================
# 2. TRANSFORMING ORDINAL FEATURES
# 'Generation' has a natural order (Gen 1 < Gen 2 < ... < Gen 6),
# so we map it to integers instead of treating it as unordered text.
# ===================================================================
generation_map = {
    'Gen 1': 1,
    'Gen 2': 2,
    'Gen 3': 3,
    'Gen 4': 4,
    'Gen 5': 5,
    'Gen 6': 6
}
df['Generation_Ordinal'] = df['Generation'].map(generation_map)

print("\nAfter transforming 'Generation' into an ordinal feature:")
display(df[['Name', 'Generation', 'Generation_Ordinal']].head(10))


# ===================================================================
# 3. ONE-HOT ENCODING
# 'Type 1' is a categorical (nominal) feature with no natural order,
# so we one-hot encode it into separate binary columns.
# ===================================================================
df_encoded = pd.get_dummies(df, columns=['Type 1'], prefix='Type1')

print("\nAfter one-hot encoding 'Type 1' (first 10 rows, relevant columns):")
type1_cols = [c for c in df_encoded.columns if c.startswith('Type1_')]
display(df_encoded[['Name'] + type1_cols].head(10))


# ===================================================================
# Final combined dataframe with all new features
# ===================================================================
print("\nFinal dataframe with interaction, ordinal, and one-hot columns:")
display(df_encoded.head(10))

# Optional: save result
df_encoded.to_csv('Pokemon_transformed.csv', index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from IPython.display import display

# ===================================================================
# PART 1: song_views.csv  ->  numerical column: listen_count
# ===================================================================
songs = pd.read_csv('song_views.csv')

print("song_views.csv - original data (first 10 rows):")
display(songs.head(10))

# ---- Standard Scaling (mean=0, std=1) ----
std_scaler = StandardScaler()
songs['listen_count_standard_scaled'] = std_scaler.fit_transform(songs[['listen_count']])

# ---- Min-Max Scaling (range 0 to 1) ----
minmax_scaler = MinMaxScaler()
songs['listen_count_minmax_scaled'] = minmax_scaler.fit_transform(songs[['listen_count']])

print("\nsong_views.csv - after Standard and Min-Max scaling (first 10 rows):")
display(songs[['title', 'listen_count',
               'listen_count_standard_scaled',
               'listen_count_minmax_scaled']].head(10))

songs.to_csv('song_views_scaled.csv', index=False)


items = pd.read_csv('item_popularity.csv')

print("\nitem_popularity.csv - original data:")
display(items)

# ---- Standard Scaling (mean=0, std=1) ----
items['pop_percent_standard_scaled'] = std_scaler.fit_transform(items[['pop_percent']])

# ---- Min-Max Scaling (range 0 to 1) ----
items['pop_percent_minmax_scaled'] = minmax_scaler.fit_transform(items[['pop_percent']])

print("\nitem_popularity.csv - after Standard and Min-Max scaling:")
display(items)

items.to_csv('item_popularity_scaled.csv', index=False)
```

```python
# ===================================================================
# Fixed-Width Binning, Adaptive (Quantile) Binning, and Log Transform
# Dataset: fcc_2016_coder_survey_subset.csv
# ===================================================================

import pandas as pd
import numpy as np
from IPython.display import display

df = pd.read_csv('fcc_2016_coder_survey_subset.csv')

print("Original data (first 10 rows):")
display(df[['ID.x', 'Age', 'Income']].head(10))


# ===================================================================
# 1. FIXED-WIDTH BINNING  ->  on 'Age'
# Split Age into equal-width bins (e.g. every 10 years) with labels.
# ===================================================================
bin_edges = [0, 20, 30, 40, 50, 60, 100]
bin_labels = ['<20', '20-29', '30-39', '40-49', '50-59', '60+']

df['Age_fixed_bin'] = pd.cut(df['Age'], bins=bin_edges, labels=bin_labels, right=False)

print("\nAfter fixed-width binning on 'Age':")
display(df[['ID.x', 'Age', 'Age_fixed_bin']].head(10))

print("\nCount of respondents per fixed Age bin:")
display(df['Age_fixed_bin'].value_counts().sort_index())


# ===================================================================
# 2. ADAPTIVE BINNING (QUANTILE-BASED)  ->  on 'Income'
# Split Income into bins that each contain roughly the same number
# of observations, using quantiles instead of fixed cutoffs.
# ===================================================================
df['Income_adaptive_bin'] = pd.qcut(
    df['Income'],
    q=4,
    labels=['Low', 'Medium', 'High', 'Very High']
)

print("\nAfter adaptive (quantile) binning on 'Income':")
display(df[['ID.x', 'Income', 'Income_adaptive_bin']].head(10))

print("\nCount of respondents per adaptive Income bin:")
display(df['Income_adaptive_bin'].value_counts().sort_index())


# ===================================================================
# 3. LOG TRANSFORM  ->  on 'Income'
# Income is highly skewed (a few very high earners), so a log
# transform compresses the scale and makes the distribution more
# symmetric. We use log1p (log(1 + x)) to safely handle any zeros.
# ===================================================================
df['Income_log'] = np.log1p(df['Income'])

print("\nAfter log transform on 'Income':")
display(df[['ID.x', 'Income', 'Income_log']].head(10))

print("\nIncome skewness before vs after log transform:")
print("Original Income skew:", df['Income'].skew())
print("Log-transformed Income skew:", df['Income_log'].skew())


# ===================================================================
# Save the final result
# ===================================================================
df.to_csv('fcc_survey_transformed.csv', index=False)
```

```python
# ===================================================================
# Bag of Words - Bigram (2-gram) Model
# Dataset: const.txt (the U.S. Constitution)
# ===================================================================

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from IPython.display import display

# ---- Step 1: Load the text file ----
with open('const.txt', 'r', encoding='utf-8', errors='ignore') as f:
    text = f.read()

# Split the document into "sentences" / chunks so BoW has multiple
# rows to compare (splitting on blank lines/paragraphs works well
# for a structured document like the Constitution).
documents = [p.strip().replace('\n', ' ') for p in text.split('\n\n') if p.strip()]

print(f"Number of text chunks (documents): {len(documents)}")
print("\nFirst 3 chunks:")
for doc in documents[:3]:
    print("-", doc[:150], "...\n")


# ===================================================================
# Step 2: Build the Bag of Words model using bigrams (2-grams)
# ngram_range=(2, 2) means only 2-word phrases are counted
# ===================================================================
vectorizer = CountVectorizer(ngram_range=(2, 2), stop_words='english')
bow_matrix = vectorizer.fit_transform(documents)

# Convert to a DataFrame for easy viewing
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\nBag of Words (Bigram) matrix shape:", bow_df.shape)
print("\nBag of Words (Bigram) table (first 10 rows, first 15 bigram columns):")
display(bow_df.iloc[:10, :15])


# ===================================================================
# Step 3: Most frequent bigrams across the whole document
# ===================================================================
bigram_totals = bow_df.sum(axis=0).sort_values(ascending=False)

print("\nTop 20 most frequent bigrams in the Constitution:")
display(bigram_totals.head(20))


# ===================================================================
# Optional: save results
# ===================================================================
bow_df.to_csv('const_bow_bigrams.csv', index=False)
bigram_totals.to_csv('const_top_bigrams.csv', header=['count'])
```
